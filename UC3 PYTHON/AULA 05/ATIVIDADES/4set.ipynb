# 1. O que é um Conjunto(set)?


### O que é?

Um set é uma coleção que guarda itens sem repetir nenhum.

jogos = {"GTA V", "Forza", "Minecraft", "GTA V"}print(jogos)



O Python guarda "GTA V" apenas uma vez, mesmo que você tenha escrito duas vezes.
Características

- Não ordenado: você não deve confiar na posição dos itens.
- Sem duplicatas: não guarda valores repetidos.
- Mutável: você pode adicionar e remover itens.
- Itens imutáveis: os elementos precisam ser de tipos que possam ser usados em conjuntos, como textos e números.
  
Comandos importantes:
add() - adicionar

jogos.add("Minecraft")



Adiciona "Minecraft" ao conjunto, caso ainda não exista.

remove() - remover
jogos.remove("Forza")



Remove "Forza". Se ele não existir, ocorre um erro.

in — verificar se existe
print("GTA V" in jogos)



Resultado: True, porque "GTA V" está no conjunto.

### Quando usar?

Quando você quer eliminar repetições ou verificar rapidamente se um item existe.

### Atenção: um conjunto vazio é criado com set(), e não com {}, porque {} cria um dicionário vazio.
