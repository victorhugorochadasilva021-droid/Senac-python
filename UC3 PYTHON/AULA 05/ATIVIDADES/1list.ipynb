# 1. O que é uma lista (list)?

### É como uma caixa onde você guarda várias informações.
projetos = ["Vendas", "Usuários", "Relatórios"]

# 2. Características das listas

### 1. Ordenada

Os itens mantêm a ordem em que foram colocados.
frutas = ["Maçã", "Banana", "Uva"]



A maçã continua sendo o primeiro item.

### 2. Mutável

frutas[0] = "Laranja"

### 3. Permite duplicatas

Você pode repetir o mesmo item várias vezes.
frutas = ["Maçã", "Maçã", "Uva"]

### 4. Heterogênea

Você pode guardar diferentes tipos de informação na mesma lista.
dados = ["Victor", 17, 1.88, True]

- "Victor": texto.
- 17: número inteiro (int).
- 1.88: número decimal (float).
- True: valor verdadeiro (bool).
  

# 3. Quando usar listas?
- Guardar vários dados: nomes, produtos, projetos etc.
- Adicionar ou remover itens: quando sua lista precisa mudar.
- Criar pilhas ou filas: estruturas que organizam a entrada e a saída de dados.

# 4. Entendendo o código do seu exemplo

### 
CRIANDO A LISTA
projetos = ["Vendas", "Usuários", "Relatórios"]



Você criou uma lista com três projetos.

ADICIONANDO UM ITEM
projetos.append("Dashboard")



append() adiciona um novo item ao final da lista.

MOSTRANDO A LISTA
print(projetos)



print() mostra o conteúdo na tela.

ACESSANDO UM ITEM
primeiro = projetos[0]



Os índices começam em zero. Então, [0] pega o primeiro item da lista.

MOSTRANDO UMA MENSAGEM
print(f"Primeiro projeto: {primeiro}")



O f permite colocar o valor de uma variável dentro do texto usando {}.
