# 1. O que é uma tupla (tuple)?

### Uma tupla é parecida com uma lista: ela guarda vários itens em uma única variável.

A principal diferença é que uma tupla não pode ser modificada depois de criada.

Jogos = ("GTA V", "Minecraft", "Forza)

Aqui, guardamos três jogos dentro da tupla jogos.
- Listas usam colchetes [].
- Tuplas usam parênteses ().


# 2. Características das tuplas

### 1. Ordenada

Os itens mantêm a ordem em que foram colocados.

cores = ("Azul", "Preto", "Branco")

O primeiro item continua sendo "Azul".

### 2. Imutável

Você não pode alterar, adicionar ou remover itens diretamente depois de criar a tupla.

cores = ("Azul", "Preto", "Branco") 

### Isso daria erro: cores[0] = "Vermelho"

O Python apresenta um erro porque você tentou modificar a tupla.

### 3. Permite duplicatas

Você pode repetir o mesmo item.

numeros = (10, 10, 20, 30)



Isso funciona normalmente.

### 4. Heterogênea

Pode guardar diferentes tipos de dados na mesma tupla.
dados = ("Victor", 17, 1.88, True)



- "Victor": texto.
- 17: número inteiro.
- 1.88: número decimal.
- True: verdadeiro.

# 3. Quando usar tuplas?

- Dados que não devem mudar: por exemplo, coordenadas de um lugar.
- Retornar vários valores de uma função: uma função pode devolver dois ou mais resultados juntos.
- Chaves de dicionários: uma tupla formada por valores imutáveis pode ser usada como chave, desde que todos os seus elementos também sejam utilizáveis como chaves.

# 4. Entendendo um exemplo

### coordenadas = (-22.90, -43.20)

Guardamos duas coordenadas em uma tupla.

ACESSANDO UM ITEM

print(coordenadas[0])



Mostra -22.90, porque o índice 0 representa o primeiro item.

CONTANDO OS ITENS

print(len(coordenadas))



Mostra 2, porque a tupla contém dois itens.

# 5. Resumo para memorizar

### Conceito	Significado simples

tuple =	Coleção de itens que não podem ser modificados

() = Parênteses usados para criar uma tupla

Ordenada	= Mantém a ordem dos itens

Imutável	= Não permite alterar os itens diretamente

len()	= Conta quantos itens existem

[0]	= Acessa o primeiro item

### A diferença mais importante:

### Lista: pode mudar 

lista = ["GTA V", "Forza"]lista.append("Minecraft")

Tupla: não pode mudar diretamente

tupla = ("GTA V", "Forza")



Pensa assim: lista é um caderno em que você pode apagar e escrever; tupla é uma informação que você guarda sem poder alterar seus itens.
