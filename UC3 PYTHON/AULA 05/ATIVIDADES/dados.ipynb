# 1. O que é persistência de dados

Até agora, você aprendeu a guardar informações em variáveis, listas, tuplas e dicionários.

Mas existe um problema: quando o programa termina, esses dados normalmente desaparecem da memória.

A persistência serve para salvar os dados e recuperá-los depois, mesmo que você feche o programa ou desligue o PC.

Por exemplo, imagine um sistema de cadastro de clientes. Você não quer perder todos os clientes quando fechar o programa, certo?

Existem diferentes formas de resolver isso.

# 2. Arquivos de texto (.txt)

São arquivos comuns que guardam texto, como os que você abre no Bloco de Notas.
Comandos principais
- open(): abre ou cria um arquivo.
- write(): escreve no arquivo.
- read(): lê o conteúdo do arquivo.
- with: organiza a abertura e o fechamento do arquivo automaticamente.
- 
Exemplo:

dados = "Meu primeiro arquivo Python!"with open("dados.txt", "w") as arquivo:    arquivo.write(dados)



O que cada parte faz?
- "dados.txt": nome do arquivo.
- "w": modo de escrita. Cria o arquivo ou substitui seu conteúdo se ele já existir.
- as arquivo: dá o nome arquivo ao arquivo aberto.
- arquivo.write(dados): escreve o conteúdo dentro dele.
Para ler o arquivo:
with open("dados.txt", "r") as arquivo:    

conteudo = arquivo.read()print(conteudo)



O "r" significa leitura.

Importante: "w" apaga o conteúdo anterior quando o arquivo já existe. Para acrescentar conteúdo sem apagar o anterior, existe o modo "a".

# 3. Arquivos CSV (.csv)

CSV é um formato que organiza informações em linhas e colunas, como uma planilha do Excel.

Por exemplo:
nome,idade

Victor,17

Rebeka,18


Isso representa uma tabela:

nome	idade

Victor	17

Rebeka 18

O Python possui um módulo chamado csv para trabalhar com esse formato.

Quando usar? Quando você precisa salvar listas de produtos, cadastros ou outros dados organizados em tabelas.

# 4. JSON (.json)

JSON é um formato muito utilizado para guardar e trocar informações entre programas e sites.

Ele se parece bastante com um dicionário Python.

Exemplo:

import jsondados = {    "nome": "Victor",    "jogo": "GTA V"}with open("dados.json", "w") as arquivo:    json.dump(dados, arquivo, indent=4)



Vamos entender:
- import json: carrega o módulo que trabalha com JSON.
- dados: é o dicionário que queremos salvar.
- json.dump(): grava os dados no arquivo.
- indent=4: organiza o conteúdo para ficar mais fácil de ler.
Para recuperar os dados:

with open("dados.json", "r") as arquivo:    dados_lidos = json.load(arquivo)print(dados_lidos["nome"])



Resultado: Victor.

Duas palavras importantes
- Serialização: transformar os dados Python em um formato que possa ser salvo.
- Desserialização: transformar os dados salvos novamente em estruturas Python.
Pensa assim: serializar é guardar; desserializar é recuperar.

# 5. Pickle (.pkl)

O pickle também serve para salvar e recuperar dados Python.

A diferença é que ele consegue preservar muitos tipos de objetos Python que o JSON não consegue representar diretamente.

Exemplo:

import pickledados = {"nome": "Victor", "idade": 17}with open("dados.pkl", "wb") as arquivo:    pickle.dump(dados, arquivo)



Aqui:
- import pickle: importa o módulo.
- "wb": abre o arquivo para escrita em formato binário.
- pickle.dump(): salva os dados.
Para recuperar:
with open("dados.pkl", "rb") as arquivo:    dados_lidos = pickle.load(arquivo)



- "rb": abre o arquivo para leitura em formato binário.
- pickle.load(): recupera os dados.
- 
Cuidado: nunca use pickle.load() em arquivos de fontes desconhecidas ou não confiáveis. Um arquivo malicioso pode executar código durante a leitura.

Para começar, você pode priorizar o JSON, que é mais fácil de visualizar e compartilhar.

# 6. Banco de dados SQLite (.db)

### Um banco de dados é uma forma de armazenar e organizar informações para que você possa pesquisar, adicionar, alterar e remover dados.

O SQLite guarda essas informações em um arquivo, sem precisar manter um servidor de banco de dados separado.

É muito útil em projetos Python.

### Imagine uma tabela de desenvolvedores:

id	nome	linguagem

1	Victor	Python

2	Rebeka	JavaScript

### Comandos principais:

connect() — conectar

conexao = sqlite3.connect("dados.db")



### Abre o banco ou cria o arquivo se ele ainda não existir.

cursor() — preparar consultas

cursor = conexao.cursor()



### Cria um objeto que permite executar comandos SQL.

execute() — executar um comando

cursor.execute(    "SELECT * FROM desenvolvedores")



### Executa uma consulta SQL para buscar os registros da tabela.

commit() — salvar alterações

conexao.commit()



### Confirma as alterações feitas no banco.

fetchall() — pegar os resultados
registros = cursor.fetchall()



### Recupera todas as linhas retornadas pela consulta.

close() — fechar a conexão
conexao.close()
