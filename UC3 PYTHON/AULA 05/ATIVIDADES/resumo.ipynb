# 1. Resumo final para estudar

### Técnica	Para que serve?
set	Guardar itens sem repetições

.txt	Guardar texto simples

CSV	Guardar dados em formato de tabela

JSON	Salvar e compartilhar dados estruturados

Pickle	Salvar objetos Python em formato binário

SQLite	Armazenar e consultar dados em tabelas

Minha sugestão para você, que está aprendendo Python e desenvolvimento de software no SENAC, é estudar nesta ordem:

1. set e suas operações.
2. Arquivos .txt com open(), read() e write().
3. JSON com json.dump() e json.load().
4. CSV.
5. SQLite e comandos SQL.
6. Pickle, quando precisar trabalhar com objetos Python mais complexos.

### Você não precisa memorizar todos os comandos de uma vez. Primeiro entenda para que cada ferramenta serve; depois pratique com pequenos programas.
