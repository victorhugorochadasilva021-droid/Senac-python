# 1. O que é um dicionário (dict)?

Um dicionário guarda informações em pares de chave e valor.

Pensa em uma ficha de cadastro: você tem o campo "nome" e o valor "Victor".

### Nesse exemplo:
- "nome" é a chave.
- "Victor" é o valor.
- "idade" é a chave.
- 17 é o valor.
- 
É como se cada informação tivesse uma etiqueta para você encontrar o que precisa.

# 2. Características dos dicionários

### 1. Chave e valor

Cada chave identifica uma informação.

carro = {    "marca": "BMW",    "ano": 2025}



A chave "marca" guarda o valor "BMW".

Para acessar esse valor:

print(carro["marca"])



Resultado: BMW.

### 2. Mutável

Você pode adicionar, modificar e remover informações.

carro["cor"] = "Preto"

Adicionamos a informação "cor": "Preto" ao dicionário.

Para modificar uma informação existente:

carro["ano"] = 2026



Agora, o ano é 2026.

### 3. Chaves únicas

As chaves não devem se repetir. Se você atribuir um novo valor a uma chave existente, o valor anterior será substituído.

pessoa = {    "nome": "Victor",    "nome": "Pedro"}



Nesse caso, "nome" terá o valor "Pedro".

### 4. As chaves precisam ser imutáveis

As chaves geralmente são textos (str) ou números (int), mas também podem ser tuplas que contenham apenas elementos utilizáveis como chaves.

dados = {    "nome": "Victor",     idade = 17}



### Atenção: esse último exemplo está escrito incorretamente. A forma certa é:

dados = {    "nome": "Victor",    "idade": 17}



As chaves são escritas antes dos dois-pontos : e os valores depois deles.

# 5. Resumo para memorizar

Conceito =	Significado simples

dict =	Guarda informações em pares

Chave =	Nome usado para encontrar uma informação

Valor =	Informação guardada

: =	Separa a chave do valor

Mutável =	Permite adicionar, alterar e remover

Chaves únicas =	Não devem se repetir

pessoa["nome"]	= Acessa o valor da chave "nome"

pop() =	Remove uma chave e seu valor


Diferença entre lista, tupla e dicionário:

- Lista []: guarda itens que você pode modificar.
- Tupla (): guarda itens que não pode modificar diretamente.
- Dicionário {}: guarda informações usando chaves e valores.
