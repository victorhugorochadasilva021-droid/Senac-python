## Soma de Elementos

```python
numeros = []

for i in range(5):
    num = int(input(f"Digite o {i+1}º número inteiro: "))
    numeros.append(num)

soma = sum(numeros)
print (f"Lista digitada: {numeros}")
print (f"Soma dos elementos: {soma}")
```

## Inserção Ordenada (sem usar .sort())

```python
numeros_ordenados = []

for i in range(5):
    num = int(input(f"Digite o {i+1}º número inteiro: "))

    inserido = False
    for j in range(len(numeros_ordenados)):
        if num < numeros_ordenados[j]:
            numeros_ordenados.insert(j, num)
            inserido = True
            break

    if not inserido:
        numeros_ordenados.append(num)

print (f"Lista ordenada: {numeros_ordenados}")
```

## Contagem de Ocorrências

```python
import random

tupla_aleatoria = tuple(random.randint(1, 5) for _ in range(10))

print (f"Tupla gerada: {tupla_aleatoria}")
busca = int(input("Digite um número entre 1 a 5 para contar: "))

quantidade = tupla_aleatoria.count(busca)
print (f"O número {busca} aperece {quantidade} vez(es) na tupla. ")
```

## Remoção de Duplicatas

```python
numeros = []
for i in range(10):
    num  = int(input(f"Digite o {i+1}º número inteiro: "))
    numeros.append(num)

unicos = []
for num in numeros:
    if num not in unicos:
        unicos.append(num)

print (f"Lista original: {numeros}")
print (f"Lista sem duplicatas: {unicos}")
```

## Estatísticas de uma Lista

```python
import statistics

numeros = []
for i in range(10):
    num = int(input(f"Digite o {i+1}º número inteiro: "))
    numeros.append(num)

maior = max(numeros)
menor = min(numeros)
media = sum(numeros)
mediana = statistics.median(numeros)

print (f"\nLista: {numeros}")
print (f"Maior valor: {maior}")
print (f"Menor valor: {menor}")
print (f"Média: {media:.2f}")
print (f"Mediana: {mediana}")
```

## Inversão de Lista

```python
frase = input("Digite uma frase: ")

palavras = frase.split()
palavras_invertidas = palavras[::-1]
resultado = " ".join(palavras_invertidas)

print (f"Saída {resultado}")
```

## Interseção de Listas

```python
import random

lista1 = [random.randint(1, 10) for _ in range (5)]

lista2 = [random.randint(1, 10) for _ in range (5)]

intersecao = []
for num in lista1:
    if num in lista2 and num not in intersecao:
        intersecao.append(num)
        
print ("--- PRIMEIRA LISTA ---")
print (f"Lista 1: {lista1}\n")

print ("--- SEGUNDA LISTA ---")
print (f"Lista 2: {lista2}\n")

print ("--- ELEMENTOS EM AMBAS AS LISTAS ---")
print (f"Elementos em ambas as listas: {intersecao}")
```

## Índice de Múltiplos de 3

```python
numeros = list(range(1, 51))

print (f"Índices e valores múltiplos de 3:")
for indice, valor in enumerate(numeros):
    if valor % 3 == 0:
        print (f"Índice {indice}: valor {valor}")
```
