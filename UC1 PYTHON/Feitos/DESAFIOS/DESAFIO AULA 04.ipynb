# Desafio 7 - Calculadora de IMC

```python
peso = float(input("Digite seu peso (kg):"))
altura = float(input("Digite a sua altura (m):"))

imc = peso / (altura ** 2)
```

```python
print (f"Seu IMC é {imc:.2f}")

if imc < 18.5:
    print ("Você está abaixo do seu peso ideal.")
elif 18.5 <= imc <= 24.9:
    print ("Você está com o peso ideal.")
else:
    print ("Você está acima do peso ideal.")
```

# Desafio 10 - DESAFIO FINAL: Simulador de Descontos

```python
valor_compra = float(input("Digite o valor total da compra: R$ "))
```

```python
if valor_compra > 500:
    desconto = 0.15
elif 200 <= valor_compra <= 500:
    desconto = 0.10
else:
    desconto = 0.05
```

```python
valor_final = valor_compra * (1 - desconto)
percentual = int (desconto * 100)

print (f"Desconto aplicado: {percentual}%")
print (f"Valor final com desconto: R$ {valor_final:.2f}")
```

# Desafio Extra - Classificação do Clima e Dica de Roupa

```python
temp = float(input("Digite a temperatura em graus Celsius: "))
```

```python
if temp < 0:
    print ("Clima: Congelante! ")
    print ("Dica: Use roupas quentes e se mantenha aquecido! ")
elif 0 <= temp <= 20:
    print ("Clima: Agradável / Fresco. ")
    print ("Dica: Use um casaco leve ou um suéter resolve bem! ")
elif 20 < temp <= 30:
    print ("Clima: Agradável / Ensolarado. ")
    print ("Dica: Use roupas de dia a dia e se mantenha hidratado! ")
else:
    print ("Clima: MUITO quente! ")
    print ("Dica: Beba bastante água, use roupas leves e evite exposição prolongada ao sol! ")
```
