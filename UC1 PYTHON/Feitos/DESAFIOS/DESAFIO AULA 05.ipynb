## Jogo da Advinhação


```python
numero_secreto = 7
acertou = False

for tentativa in range(1, 4):
    chute = int(input(f"Tentativa {tentativa}/3 - Advinhe o número (1 a 10): "))
    if chute == numero_secreto:
        print ("Parabéns, você acertou!")
        acertou=True
        break
if not acertou:
    print ("Você perdeu! O nùmero era", numero_secreto)
```

## Par ou Ímpar e Positivo ou Negativo?

```python
num = int(input("Digite um número: "))
paridade = "Par" if num % 2 == 0 else "Ímpar"
if num > 0:
    sinal = "Positivo"
elif num < 0:
    sinal = "Negativo"
else:
    sinal = "Neutro (Zero)"
print (f"O número é {paridade} e {sinal}.")
```

## Qual Tipo de Triângulo

```python
l1 = float(input("Digite o primeiro lado: "))
l2 = float(input("Digite o segundo lado: "))
l3 = float(input("Digite o terceiro lado: "))
```

```python
if l1 == l2 == l3:
    print ("Triângulo Equilátero (todos os lados iguais)")
elif l1 == l2 or l1 == l3 or l2 == l3:
    print ("Triângulo Isósceles (dois lados iguais)")
else:
    print ("Triângulo Escaleno (todos os lados diferentes)")
```
