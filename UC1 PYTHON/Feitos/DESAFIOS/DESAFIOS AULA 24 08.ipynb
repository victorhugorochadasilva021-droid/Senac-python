## Calcular o fatorial de um número usando while

```python
num = int(input("Digite um número: "))
fatorial = 1
temp = num
```

```python
while temp > 1:
    fatorial *= temp
    temp -= 1
```

```python
print (f"O fatorial de {num} é: {fatorial} ")
```

## Números primos

```python
num = int(input("Digite um número inteiro positivo: "))
eh_primo = True
```

```python
if num <= 1:
    eh_prime = False
else: 
    for i in range (2, int(num**0.5) + 1):
        if num % i == 0:
            eh_primo = False
            break

if eh_primo:
    print (f"O número {num} é primo.")
else:
    print (f"O número {num} NÃO é primo. ")
```

## Sequência de Fibonacci (10 primeiros números com while)

```python
a, b = 0, 1
contador = 0
```

```python
while contador < 10:
    print (a, end=" " if contador < 9 else "\n")
    a, b = b, a + b
    contador += 1
```

## Menu interativo

```python
opcao = ""
```

```python
while opcao.lower() != "sair":
    print ("\n--- MENU ---")
    print ("1. Olá")
    print ("2. Ajuda")
    print ("Digite 'sair' para encerrar")

    opcao = input ("Escolha uma opção: ")

    if opcao == "1":
        print ("Olá! Tudo bem?")
    elif opcao == "2":
        print (" Esta é a seção de ajuda. ")
    elif opcao.lower() != "sair":
        print ("Opção inválida!")

print ("Programa encerrado. ")
```
