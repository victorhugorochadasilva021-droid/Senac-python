```python
try:
    
    numero = int(input("Digite um número inteiro: "))

    if numero % 2 == 0:
        print (f"O número {numero} é PAR.")
    else:
        print (f"O número {numero} é IMPAR.")
except ValueError:
    print ("Entrada inválida. Por favor, digite apenas números inteiros. ")
```

# Desafio Aula 04.2

```python

peso = float(input("Digite o seu peso em kg (ex: 70.5): "))
altura = float(input("Digite a sua altura em metros (ex: 1.75): "))

imc = peso / (altura ** 2)
if imc <= 18.5:
    classificacao = "Abaixo do peso. É necessário uma alimentação calórica e saudável."
elif imc <= 24.9:
    classificacao = "Peso ideal. Sem preocupações."
elif imc <= 29.9:
    classificacao = "Sobrepeso. É necessário uma dieta controlada."
else:
    classificacao = "Obesidade. É necessário dieta controlada e supervisionada, junto de exercícios fisícos."        
```

```python
print(f"\nSeu IMC  é: {imc: .2f}")
print(f"Classificação: {classificacao}")
```

# Gerar todas as combinações de A e B

```python
for  A in [0,1]:
    for B in [0,1]:
        resultado = A and B # Operação AND
        print (f" {A} | {B} | {resultado} ")
```
