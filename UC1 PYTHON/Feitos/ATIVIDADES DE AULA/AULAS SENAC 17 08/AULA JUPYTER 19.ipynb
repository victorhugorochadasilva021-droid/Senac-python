### Desafio 3.8

```python
capital = float(input("Digite o valor inicial do investimento: R$ "))
taxa = float(input("Digite a taxa de juros mensal (em %): ")) / 100
tempo_meses = int(input("Digite o tempo em meses: "))
```

```python
juros = capital * taxa * tempo_meses
montante = capital + juros
```

```python
print (f"O montante final após {tempo_meses} meses é: R$ {montante: .2f}")
```

### Desafio 3.9

```python
a = float(input("Digite o valor de a: "))
b = float(input("Digite o valor de b: "))
c = float(input("Digite o valor de c: "))

delta = (b ** 2) - (4 * a * c)


```

```python
if delta < 0 or a == 0:
    print ("A equação não possui raízes reais.")
else:
    x1 = (-b + (delta ** 0.5)) / (2 * a)
    x2 = (-b (delta ** 0.5)) / (2 * a)
    print (f"As raízes são : x1 = {x1:.2f}e x2 = {x2: .2f}")
```
