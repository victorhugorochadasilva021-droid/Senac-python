### Função par_ou_impar

```python
def par_ou_impar(n):
    if n % 2 == 0:
        return "Par"
    else:
        return "Ímpar"
    
# Exemplo de chamada
print ("O número 4 é", par_ou_impar(4))
print ("O número 7 é", par_ou_impar(7))
```

### Função fatorial

```python
def fatorial (n):
    if n < 0:
        return "Número inválido para fatorial. "
    elif n == 0:
        return 1
    else:
        resultado = 1
        for i in range (1, n + 1):
            resultado *= i
        return resultado
# Exemplo de chamada
print ("Fatorial de 5:", fatorial(5))
```

### Função tabuada

```python
def tabuada (n):
    resultado = []
    for i in range (1, 11):
        resultado.append (n * i)
    return resultado

# Exemplo de chamada
print ("Tabuada do 5:", tabuada(5))
```

### Função ordenar_por_segundo

```python
def ordenar_por_segundo (lista_tuplas):
    return sorted (lista_tuplas, key =lambda x:x[1])

# Exemplo de chamada
lista_exemplos = [("a", 3), ("b", 1), ("c", 2)]
print ("Lista Ordenada: ", ordenar_por_segundo (lista_exemplos))
```

## Escreva um programa Python que use a função main() para calcular o fatorial de um número fornecido pelo usuário

```python
def fatorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fatorial (n - 1)

def main ():
    num = int(input("Digite um número: "))
    print (f"O fatorial de {num} é {fatorial(num)}")

if __name__ == "__main__":
    main ()
```

## Desenvolva um programa Python que use a função main() para verificar se um número é primo ou não


```python
def eh_primo(n):
    if n <= 1:
        return False
    for i in range (2, int(n**0.5) + 1):
        if n % i == 0:
            return False
        return True

def main():
    num = int(input("Digite um número: "))
    if eh_primo (num):
        print (f"{num} é primo. ")
    else:
        print (f"{num} não é primo. ")

if __name__ == "__main__":
    main()
        
```

## Escreva um programa Python que use a função main() para contar quantas vogais existem em uma string fornecida pelo usuário

```python
def contar_vogais(texto):
    vogais = "aeiouAEIOU"
    return sum (1 for char in texto if char in vogais)

def main():
    texto = input("Digite uma string: ")
    print (f"A string contém {contar_vogais(texto)} vogais.")

if __name__ == "__main__":
    main()
```

## Desenvolva um programa Python que use a função main() para gerar a sequência de Fibonacci até um número fornecido pelo usuário.

```python
def fibonacci(n):
   sequencia = []
   a, b = 0, 1
   while a <= n:
        sequencia.append(a)
        a, b = b, a + b
   return sequencia
```

```python
def main():
   num = int(input("Digite um número: "))
   print (f"Sequência de Fibonacci até {num}: {fibonacci(num)}")
```

```python
if __name__ == "__main__":
   main()
```
