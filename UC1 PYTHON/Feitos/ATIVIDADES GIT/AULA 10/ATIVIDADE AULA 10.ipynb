### Soma dos Elementos da Diagonal Principal

```python
import random

matriza = []
for i in range (4):
    linha = []
    for j in range(4):
        linha.append(random.randint(1, 99))
    matriz.append(linha)
    
print ("Matriz 4x4 gerada: ")
for linha in matriz:
    for elemento in linha:
        print (f"{elemento:4}", end=" ")
    print()

soma_diagonal = 0
for i in range(4):
    soma_diagonal += matriz[i][i]

print (f"\nSoma da diagonal principal: {soma_diagonal}")
```

### Soma dos Elementos da Diagonal Secundária

```python
import random

matriz = []
for i in range(4):
    linha = []
    for j in range(4):
        linha.append(random.randint(1, 100))
    matriz.append(linha)

print ("Matriz 4x4 gerada: ")
for linha in matriz:
    for elemento in linha:
        print (f"{elemento:4}", end= " ")
    print()

soma_secundaria = 0
for i in range(4):
    soma_secundaria += matriz [i][3 - i]

print (f"\nSoma de diagonal secundária: {soma_secundaria}")
```

### Contagem de Números Pares

```python
matriza = []
pares = 0
impares = 0

print ("Digite os elementos da matriz 4x4:")

for i in range(4):
    linha = []
    for j in range(4):
        while True:
            try:
                elemento = int(input(f"Elemento [{i}][{j}]: "))
                linha.append (elemento)

                if elemento % 2 == 0:
                    pares += 1
                else:
                     impares += 1
                break
            except ValueError:
                 print ("Por favor, digite um número inteiro válido.")
    matriz.append(linha)

print ("\nMatriz Informada:")
for linha in matriz:
    for elemento in linha:
        print (f"{elemento:5}", end=" ")
    print ()

print (f"\nQuantidade de números pares: {pares}")
print (f"Quantidade de números ímpares: {impares}")
```

### Multiplicação de Linhas por um Número

```python
import random

def exibir_matriz(matriz, mensagem):
    print (f"\n{mensagem}:")
    for linha in matriz:
        for elemento in linha:
            print (f"{elemento:6}", end=" ")
        print()

matriz = [[random.randint(1, 20) for _ in range(4)] for _ in range(4)]

exibir_matriz(matriz,"Matriz original")

while True:
    try:
        linha = int (input("\nDigite o número da linha (0 a 3) para multiplicar: "))
        if 0 <= linha <= 3:
            break
        else:
            print ("Por favor, digite um número entre 0 e 3.")
    except ValueError:
        print ("Por favor, digite um número inteiro válido.")
while True:
    try:
        multiplicador = int(input("Digite o valor inteiro para multiplicar: "))
        break
    except ValueError:
        print ("Por favor, digite um número inteiro válido. ")
matriz[linha] = [elemento * multiplicador for elemento in matriz [linha]]

exibir_matriz(matriz, "Matriz atualizada")
```

### Multiplicação de Matrizes

```python
import random 

def criar_matriz():
    return [[random.randint(1, 10)for _ in range (4)] for _ in range(4)]

def exibir_matriz(matriz, nome):
    print (f"\n{nome}:")
    for linha in matriz:
        print (" ".join(f"{num:4}" for num in linha))

def multiplicar_matrizes(matriz_a, matriz_b):
    resultado = [[0 for _ in range (4) for _ in range (4)]]
    for i in range(4):
        for j in range(4):
            for k in range(4):
                resultado[i][j] += matriz_a[i][k] * matriz_b[k][j]
        return resultado

matriz_a = criar_matriz()
matriz_b = criar_matriz()

exibir_matriz(matriz_a, "Matriz A")
exibir_matriz(matriz_b, "Matriz B")

matriz_produto = multiplicar_matrizes(matriz_a, matriz_b)

exibir_matriz (matriz_produto, "Produto matricial A x B")
```

### Verificação de Simetria

```python
import random

def criar_matriz_aleatoria():
    """Cria uma matriz 4x4 com valores aleatórios entre 1 e 9"""
    matriz = []
    for i in range(4):
        linha = []
        for j in range(4):
            linha.append(random.randint(1, 9))
        matriz.append(linha)
    return matriz

def criar_matriz_simetrica():
    """Cria uma matriz 4x4 simétrica com valores aleatórios entre 1 e 9"""
    matriz = []
    for i in range(4):
        linha = [0] * 4
        matriz.append(linha)

    for i in range(4):
        for j in range(i, 4):
            valor = random.randint(1, 9)
            matriz[i][j] =valor
            matriz[j][i] = valor
    return matriz

def exibir_matriz(matriz, nome):
    """Exibe uma matriz formartada"""
    print(f"\n{nome}:")
    for linha in matriz:
        for numero in linha:
            print (f"{numero:3}", end=" ")
        print()

def verificar_simetria(matriz):
    """Verificar se uma matriz é simétrica"""
    for i in range(4):
        for j in range(4):
            if matriz [j][i]:
                return False
    return True

print ("Criação e análise de matrizes 4x4")

matriz_aleatoria = criar_matriz_aleatoria()
matriz_simetrica = criar_matriz_simetrica()

exibir_matriz(matriz_aleatoria, "Matriz Aleatória")
exibir_matriz(matriz_simetrica, "Matriz Simétrica")

print (" \n Resultados:")
res_aleatoria = "Sim" if verificar_simetria(matriz_aleatoria) else "Não"
print (f"A matriz aleatoria é simétrica? {res_aleatoria}")

res_aleatoria = "Sim" if verificar_simetria(matriz_simetrica) else "Não" 
print ("A matriz simétrica é simétrica? {res_simetrica}")





```

## Normalização de uma Matriz

```python
import random

def criar_matriz():
    matriz = []
    for i in range(4):
        linha = []
        for j in range(4):
            linha.append(random.randint(1, 100))
        matriz.append(linha)
    return matriz

def encontrar_maior(matriz):
    maior = matriz[0][0]
    for linha in matriz:
        for elemento in linha:
            if elemento > maior:
                maior = elemento
        return maior
    
def normalizar_matriz(matriz, maior):
    matriz_normalizada = []
    for linha in matriz:
        linha_normalizada = []
        for elemento in linha:
            linha_normalizada.append(round(elemento/maior, 4))
        matriz_normalizada.append(linha_normalizada)
    return matriz_normalizada

def exibir_matriz(matriz, titulo):
    print (f"\n{titulo}:")
    for linha in matriz:
        for elemento in linha:
            print(f"{elemento:8.4f}", end=" ")
        print ()

matriz_original = criar_matriz()
maior_valor = encontrar_maior(matriz_original)
matriz_normalizada = normalizar_matriz(matriz_original, maior_valor)

exibir_matriz(matriz_original, "Matriz Original")
print (f"\nMaior valor encontrado: {maior_valor}")
exibir_matriz(matriz_normalizada, "Matriz Normalizada")
```
